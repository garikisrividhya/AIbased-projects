```python
import time
import pandas as pd
from pymatgen.core import Composition
from matminer.featurizers.composition import ElementProperty
```

```python
# 1. Load the cleaned materials dataset from Module II
df = pd.read_csv('cleaned_cp_dataset.csv')
```

```python
# Matminer requires pymatgen Composition objects
df['composition'] = df['formula'].apply(Composition)
```

```python
# 2. Featurize with Magpie preset
magpie_feat = ElementProperty.from_preset(preset_name='magpie')
t0 = time.time()
df_magpie = magpie_feat.featurize_dataframe(df[['composition']].copy(), col_id='composition')
magpie_time = time.time() - t0
```

```python
# 3. Featurize with Deml preset
deml_feat = ElementProperty.from_preset(preset_name='deml')
t0 = time.time()
df_deml = deml_feat.featurize_dataframe(df[['composition']].copy(), col_id='composition')
deml_time = time.time() - t0
```

```python
# 4. Comparative Metrics
print("=== Featurization Benchmark ===")
print(f"Magpie features generated: {df_magpie.shape[1] - 1} | Time: {magpie_time:.2f}s")
print(f"Deml features generated:   {df_deml.shape[1] - 1} | Time: {deml_time:.2f}s")
```

```python
# 5. Export features for downstream selection
df_magpie.drop(columns=['composition']).to_csv('matminer_magpie_features.csv', index=False)
df_deml.drop(columns=['composition']).to_csv('matminer_deml_features.csv', index=False)
print("Features successfully saved for Session S20.")
```
