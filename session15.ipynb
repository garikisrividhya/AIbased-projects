```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# 1. Load the cleaned dataset containing family classifications
df = pd.read_csv('cleaned_cp_dataset.csv')
```

```python
# Ensure style settings for publication quality
sns.set_theme(style="whitegrid", font_scale=1.1)
```

```python
# 2. Create a FacetGrid for Cp vs T across chemical families
g = sns.FacetGrid(df, col="family", hue="formula", height=5, aspect=1.2, palette="tab10")
g.map(plt.plot, "T", "Cp", marker="o", linewidth=2)
g.add_legend(title="Material Formula")
```

```python
g = sns.FacetGrid(df, col="formula", hue="formula",
                  height=5, aspect=1.2, palette="tab10")

g.map(plt.plot, "T", "Cp", marker="o", linewidth=2)

g.add_legend(title="Material Formula")
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8,5))

sns.lineplot(data=df, x="T", y="Cp", hue="formula", marker="o")

plt.xlabel("Temperature (K)")
plt.ylabel("Heat Capacity Cp (J/mol-K)")
plt.title("Cp vs Temperature")
plt.show()
```
