```python
import pandas as pd
import numpy as np

# Load the saved Phase 4 datasets
X_train = pd.read_csv("X_train_filtered.csv")
X_test = pd.read_csv("X_test_filtered.csv")

y_train = pd.read_csv("y_train.csv").squeeze("columns")
y_test = pd.read_csv("y_test.csv").squeeze("columns")

# Check dataset shapes
print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)

print("\nUnique values in y_train:")
print(np.unique(y_train))
```

```python
from sklearn.svm import SVR
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
import numpy as np

# Initialize the SVR model
svr_rbf = SVR(kernel='rbf')

# Train the model
svr_rbf.fit(X_train, y_train)

print("SVR model trained successfully!")
```

```python
# Generate predictions on the test dataset
svr_preds = svr_rbf.predict(X_test)

print("SVR Predictions:")
print(svr_preds[:10])

print("\nActual Values:")
print(y_test[:10].to_numpy())
```

```python
# Evaluate the SVR model

svr_r2 = r2_score(y_test, svr_preds)

svr_rmse = np.sqrt(
    mean_squared_error(y_test, svr_preds)
)

svr_mae = mean_absolute_error(y_test, svr_preds)

print("--- SVR (RBF Kernel) Baseline Metrics ---")
print(f"Test R²:   {svr_r2:.3f}")
print(f"Test RMSE: {svr_rmse:.3f}")
print(f"Test MAE:  {svr_mae:.3f}")
```
