```python
# ============================================================
# S21 - Complete Feature Selection and Curation
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from pymatgen.core import Composition
from matminer.featurizers.composition import ElementProperty

from sklearn.feature_selection import VarianceThreshold, RFE
from sklearn.linear_model import Ridge
from sklearn.ensemble import RandomForestRegressor

# ------------------------------------------------------------
# STEP 1: Load thermodynamic dataset
# ------------------------------------------------------------

df_s21 = data.copy()

print("Original dataset shape:", df_s21.shape)

# Keep formula and target
df_s21 = df_s21[['formula_pretty', 'y']].copy()

# Remove missing values
df_s21['y'] = pd.to_numeric(df_s21['y'], errors='coerce')
df_s21 = df_s21.dropna(subset=['formula_pretty', 'y']).reset_index(drop=True)

print("Dataset after cleaning:", df_s21.shape)
print("\nTarget distribution:")
print(df_s21['y'].value_counts())


# ------------------------------------------------------------
# STEP 2: Convert formulas into Composition objects
# ------------------------------------------------------------

df_s21['composition'] = df_s21['formula_pretty'].apply(
    lambda x: Composition(x)
)

print("\nComposition conversion completed.")


# ------------------------------------------------------------
# STEP 3: Generate Magpie features
# ------------------------------------------------------------

magpie_featurizer = ElementProperty.from_preset("magpie")

df_magpie = magpie_featurizer.featurize_dataframe(
    df_s21,
    col_id='composition',
    ignore_errors=True
)

print("\nMagpie features generated.")
print("Shape after Magpie:", df_magpie.shape)


# ------------------------------------------------------------
# STEP 4: Select only numeric feature columns
# ------------------------------------------------------------

exclude_columns = [
    'formula_pretty',
    'y',
    'composition'
]

feature_columns = [
    col for col in df_magpie.columns
    if col not in exclude_columns
    and pd.api.types.is_numeric_dtype(df_magpie[col])
]

X = df_magpie[feature_columns].copy()
y = df_magpie['y'].copy()

# Replace infinity values
X = X.replace([np.inf, -np.inf], np.nan)

# Fill missing feature values with median
X = X.fillna(X.median())

print("\nInitial numeric feature count:", X.shape[1])


# ------------------------------------------------------------
# STEP 5: Variance Filter
# ------------------------------------------------------------

variance_selector = VarianceThreshold(threshold=0.01)

X_variance = variance_selector.fit_transform(X)

variance_columns = X.columns[
    variance_selector.get_support()
]

X_variance = pd.DataFrame(
    X_variance,
    columns=variance_columns,
    index=X.index
)

print(
    "Features after Variance Filter:",
    X_variance.shape[1]
)

print(
    "Dropped by Variance Filter:",
    X.shape[1] - X_variance.shape[1]
)


# ------------------------------------------------------------
# STEP 6: Correlation Filter
# ------------------------------------------------------------

corr_matrix = X_variance.corr().abs()

upper = corr_matrix.where(
    np.triu(
        np.ones(corr_matrix.shape),
        k=1
    ).astype(bool)
)

to_drop = [
    column
    for column in upper.columns
    if any(upper[column] > 0.95)
]

X_filtered = X_variance.drop(columns=to_drop)

print(
    "Features after Correlation Filter:",
    X_filtered.shape[1]
)

print(
    "Dropped by Correlation Filter:",
    len(to_drop)
)


# ------------------------------------------------------------
# STEP 7: RFE using Ridge
# ------------------------------------------------------------

ridge = Ridge(alpha=1.0)

rfe = RFE(
    estimator=ridge,
    n_features_to_select=20
)

X_rfe = rfe.fit_transform(
    X_filtered,
    y
)

top_features = X_filtered.columns[
    rfe.support_
]

print("\n======================================")
print("Top 20 Features Selected by RFE")
print("======================================")

for number, feature in enumerate(top_features, start=1):
    print(number, ":", feature)


# ------------------------------------------------------------
# STEP 8: Random Forest Feature Importance
# ------------------------------------------------------------

rf = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)

rf.fit(X_rfe, y)

importances = rf.feature_importances_

indices = np.argsort(importances)[::-1]

print("\n======================================")
print("Random Forest Feature Importance")
print("======================================")

for rank, index in enumerate(indices, start=1):
    print(
        rank,
        ":",
        top_features[index],
        "->",
        round(importances[index], 6)
    )


# ------------------------------------------------------------
# STEP 9: Plot Feature Importance
# ------------------------------------------------------------

plt.figure(figsize=(12, 6))

plt.title(
    "Random Forest Feature Importance (Top 20 RFE Features)"
)

plt.bar(
    range(len(importances)),
    importances[indices]
)

plt.xticks(
    range(len(importances)),
    top_features[indices],
    rotation=90
)

plt.xlabel("Features")
plt.ylabel("Importance")

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# STEP 10: Create Final Curated Dataset
# ------------------------------------------------------------

df_curated = df_s21[
    ['formula_pretty', 'y']
].copy()

for feature in top_features:
    df_curated[feature] = X_filtered[feature].values


# ------------------------------------------------------------
# STEP 11: Save final S21 file
# ------------------------------------------------------------

df_curated.to_csv(
    'curated_features_s21.csv',
    index=False
)

print("\n======================================")
print("✅ S21 COMPLETE")
print("======================================")

print(
    "Final curated dataset shape:",
    df_curated.shape
)

print(
    "Saved file: curated_features_s21.csv"
)

print("\nFinal columns:")
print(df_curated.columns.tolist())
```
