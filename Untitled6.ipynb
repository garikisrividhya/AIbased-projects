```python
from pymatgen.core import Composition
from pymatgen.analysis.phase_diagram import PhaseDiagram, PDEntry
```

```python
comp = Composition("NaCl")
```

```python
entries = [
    PDEntry(Composition("NaCl"), -4.0),   # formation energy in eV
    PDEntry(Composition("Na"), 0.0),
    PDEntry(Composition("Cl2"), 0.0)
]
```

```python
pd = PhaseDiagram(entries)
```

```python
stable = pd.get_decomp_and_e_above_hull(comp)
print(stable)
```
